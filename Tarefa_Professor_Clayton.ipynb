```python
# Sistema das Notas

notas = []

# Solicita a quantidade de alunos
quantidade = int(input("Digite a quantidade de alunos da sala (máximo 50): "))

# Limitação da quantidade de alunos (1 Mínimo, 50 Máximo)
if quantidade < 1 or quantidade > 50:
    print("Quantidade inválida! Digite um valor entre 1 e 50.")
else:
    for i in range(quantidade):
        nota = float(input(f"Digite a nota do aluno {i + 1}: "))

        # Validação e limitação da nota
        if nota < 0 or nota > 10:
            print("Nota inválida! Digite uma nota entre 0 e 10.")
            break

        notas.append(nota)

    # Verifica se todas as notas foram cadastradas
    if len(notas) == quantidade:

# Cálculo da média da turma
        media = sum(notas) / len(notas)

# Encontra a maior e a menor nota
maior_nota = max(notas)
menor_nota = min(notas)

# Exibição dos resultados

print("\n--- RESULTADO DA TURMA ---")
print(f"Notas Registradas: {notas}")
print(f"Média da turma: {media:.2f}")
print(f"Maior nota: {maior_nota:.2f}")
print(f"Menor nota: {menor_nota:.2f}")
```
